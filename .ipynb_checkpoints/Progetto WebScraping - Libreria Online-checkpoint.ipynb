# Progetto : Scraping delle pagine di una libreria online

L'obiettivo principale è realizzare uno script di web scraping che possa estrarre le seguenti informazioni dai libri presenti nella libreria online:

Titolo del libro
Valutazione in stelle (rating)
Prezzo del libro
Disponibilità (se è in stock o esaurito)

*   Titolo del libro
*   Valutazione in stelle (rating)
*   Prezzo del libro
*   Disponibilità (se è in stock o esaurito)


Una volta estratti, i dati devono essere organizzati in un formato strutturato e salvati in un file CSV, così da poter essere utilizzati dal team di marketing e vendita per l'analisi delle informazioni.

## Elaborazioni

Iniziamo ad impostare l'ambiente:

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

try:
  url = "https://books.toscrape.com/"
  response = requests.get(url)
  soup = BeautifulSoup(response.content, "html.parser")
  books = soup.find_all("article", class_="product_pod")
  titles = []
  rating = []
  price = []
  availability = []

except requests.exceptions.RequestException as e:
    print(f"Errore durante la richiesta HTTP: {e}")
except Exception as e:
    print(f"Si è verificato un errore: {e}")
```

Andiamo a ricercare negli oggetti i link che recupereranno il Titolo.

```python
for book in books:
  if book.find("h3").find("a"):
    titles.append(book.find("h3").find("a")["title"])                           #Accedo all'elemento 'h3' nello specifico al sotto-tag 'a' e estraggo attributo title
```

```python
titles
```

Andiamo a scaricarci la Valutazione in stelle adesso. <br>
Da una attenta valutazione del codice di ispezione emerge che la rappresentazione delle stelle di valutazione viene diversificata in un tag 'p'e nella classe utilizzata rispettivamente ad esempio per la classe con 3 stelle di valutazione viene usata la class="star-rating Three".<br><br>
Andiamo dunque a ricercare le classi del tag 'p':

```python
states = {
    "Zero": 0,
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

rating = []
for book in books:
  if book.find("p", class_="star-rating"):
    rate = book.find("p", class_="star-rating")["class"][1]                     #Accedo all'attributo di 'class' nello specifico al secondo elemento
    rating.append(states[rate])                                                 #Salvo la conversione in numero del rating

print(rating)
```

Andiamo a salvarci adesso il prezzo.<br>
Osservando l'ispezione del codice html notiamo che esso si ritrova su un tag 'p' con attributo 'price_color' nel testo in esso contenuto. Tale tags si posiziona all'interno di un tag 'div' con class 'product_price'. <br><br>
Procediamo dunque con l'estrazione:

```python
for book in books:
  if book.find("div", class_="product_price"):
    price.append(book.find("div", class_="product_price").find("p",class_="price_color").text)

price
```

Come ultimo elemento di estrazione per la compilazione del nostro db andiamo a recuperare la "Disponibilità" del libro.<br>
Anche in questo caso verificando e ispezionando il codice HTML notiamo che tale informazione è all'interno del tag 'div' con class 'product_price', all'interno del sotto-tag 'p' con class 'instock availability', e all'interno infine di un tag 'i' con class 'icon-ok'.<br><br>
Se tale tag 'i' è presente con tale classe allora possiamo dedurre che è disponibile altrimenti no.

```python
for book in books:
  if book.find("div", class_="product_price"):
    if book.find("div", class_="product_price").find("p",class_="instock availability").find("i",class_="icon-ok"):
      availability.append("In Stock")
    else:
      availability.append("Out of Stock")

availability
```

Adesso proviamo a capire come poter gestire il cambio di pagina.<br>
Per prima cosa andiamo a caricarci il totale pagine previste e la pagina corrente e ci costruiamo in metodo in grado di farlo che lo possiamo riutilizzare ad ogni iterazione.
<br>
Da un attenta valutazione del codice notiamo che sia il pulsante di "Prossima pagina" che le informazioni di avanzamento pagine sono contenute all'interno del tag 'ul' con class 'pager'. All'interno di questo tag troviamo 3 sotto-tags 'li' rispettivamente con class 'previous','current' che contiene come testo l'avanzamento e il totale pagine e infine l'ultimo tag 'li' con class 'next' che contiene un tag 'a' con url che punta alla prossima pagina.<br><br>
Procediamo dunque a caricarci l'informazione del totale pagine e la gestione della url successiva.

```python
def change_page(html_code):
    pager_ul = html_code.find("ul", class_="pager")

    if not pager_ul:
        print("Attenzione: Elemento pager non trovato.")
        return 1, 1, None                                                       # Restituisce i valori predefiniti: pagina corrente 1, totale pagine 1, nessun collegamento alla pagina successiva.

    current_li = pager_ul.find("li", class_="current")

    if not current_li:
        print("Attenzione: Elemento per lo stato della pagina corrente non trovato.")
        next_li = pager_ul.find("li", class_="next")
        next_page = next_li.find("a")["href"] if next_li and next_li.find("a") else None
        return 1, 1, next_page

    status_text = current_li.text.strip()                                       # Estrae il testo contenuto nell'elemento "current_li" e rimuove gli spazi bianchi iniziali/finali.
    current = 1                                                                 # Inizializza i valori predefiniti per la pagina corrente e totale nel caso in cui il parsing fallisca.
    total = 1

    numbers_in_status = [int(part) for part in status_text.split() if part.isdigit()]

    if len(numbers_in_status) >= 2:
        current = numbers_in_status[0]
        total = numbers_in_status[1]
    else:
        # Se non si trovano almeno due numeri nel testo, stampa un avviso.
        print(f"Attenzione: Non sono stati trovati almeno due numeri nel testo dello stato della paginazione: '{status_text}'")

    next_li = pager_ul.find("li", class_="next")
    next_page = next_li.find("a")["href"] if next_li and next_li.find("a") else None


    if next_page:
      next_page = next_page.replace("catalogue/", "") if next_page else None      #Normalizzo la URL della pagina seguente
      next_page = '/catalogue/' + next_page

    return current, total, next_page
```

```python
change_page(soup)
```

Per completezza andiamoci a creare un metodo che riceve quindi i 4 array (Title, Rating, Price, Availability) e restituisce un dataframe utile per la concatenazione dei dati tra le varie pagine e la creazione del csv in output al processo di web scraping.

```python
def create_dataframe(titles, rating, price, availability):
    data = {
        "Title": titles,
        "Rating": rating,
        "Price": price,
        "Availability": availability
    }
    df = pd.DataFrame(data)
    return df
```

Perfetto adesso che abbiamo tutti gli elementi, industrializziamo la soluzione creando un metodo che iteri l'alimentazione dei campi per ciascuna pagina fin quando non abbiamo effettuato lo scorrimento di tutte le pagine web dei book.<br>
<br>
Alla luce di questo andiamo a definire un metodo che ripercorre le estrazioni dei campi Titolo, Rating, Prezzo, Disponibilità per una pagina e lo facciamo iterare 'n' volte andando sempre su pagine diverse indicate dal metodo 'change-page'.

```python
from urllib.parse import urljoin

def extract_data(base_url):

    df_union = pd.DataFrame() # Inizializza il DataFrame vuoto per accumulare i risultati
    current_url = base_url

    states = {
        "Zero": 0,
        "One": 1,
        "Two": 2,
        "Three": 3,
        "Four": 4,
        "Five": 5
    }

    while current_url:                                                          # Continua finché c'è un URL valido per la pagina corrente o successiva

        try:
            response = requests.get(current_url)
            response.raise_for_status()                                         # Lancia un'eccezione per status code di errore (4xx o 5xx)
            soup = BeautifulSoup(response.content, "html.parser")
            books = soup.find_all("article", class_="product_pod")

            titles_page = []
            rating_page = []
            price_page = []
            availability_page = []

            for book in books:
                                                                                # Estrazione dei Titoli
                if book.find("h3") and book.find("h3").find("a"):
                    titles_page.append(book.find("h3").find("a")["title"])

                                                                                # Estrazione delle Valutazioni
                if book.find("p", class_="star-rating"):
                    rate_class = book.find("p", class_="star-rating")["class"]
                    if len(rate_class) > 1 and rate_class[1] in states:
                         rating_page.append(states[rate_class[1]])
                    else:
                        rating_page.append(None)                                # O un valore di default se il rating non è trovato/valido
                else:
                    rating_page.append(None)

                                                                                # Estrazione dei Prezzi
                if book.find("div", class_="product_price") and book.find("div", class_="product_price").find("p", class_="price_color"):
                     price_page.append(book.find("div", class_="product_price").find("p",class_="price_color").text)
                else:
                     price_page.append(None)

                                                                                # Estrazione della Disponibilità
                if book.find("div", class_="product_price") and book.find("div", class_="product_price").find("p", class_="instock availability"):
                    if book.find("div", class_="product_price").find("p",class_="instock availability").find("i",class_="icon-ok"):
                        availability_page.append("In Stock")
                    else:
                        availability_page.append("Out of Stock")                # Se l'elemento 'i' non c'è, potrebbe essere esaurito
                else:
                    availability_page.append("Unknown")


                                                                                # Crea il Dataframe dell'i-esima pagina
            df_page = create_dataframe(titles_page, rating_page, price_page, availability_page)
            df_union = pd.concat([df_union, df_page], ignore_index=True)

            curr_page, tot_page, next_page_relative = change_page(soup)         # Ottieni l'URL della pagina successiva

                                                                                # Costruisci l'URL completo per la prossima iterazione
            if next_page_relative:
                current_url = urljoin(base_url, next_page_relative)
            else:
                current_url = None                                              # Imposta a None per terminare il ciclo

            print(f"Pagina corrente: {curr_page}/{tot_page}")
            print(f"URL della pagina successiva: {next_page_relative}")
            print(f"URL della pagina corrente: {current_url}")

        except requests.exceptions.RequestException as e:
            print(f"Errore durante la richiesta HTTP per {current_url}: {e}")
            current_url = None                                                  # Ferma l'iterazione in caso di errore HTTP grave
        except Exception as e:
            print(f"Si è verificato un errore durante l'elaborazione della pagina {current_url}: {e}")
            current_url = None                                                  # Ferma l'iterazione in caso di altri errori

    return df_union                                                             # Restituisco il Dataframe al termine dell'iterazione
```

Perfetto abbiamo creato il metodo di parsing del sito, proviamo ad eseguirlo:

```python
df = extract_data('https://books.toscrape.com')
df.to_csv('books.csv', index=False)
```
